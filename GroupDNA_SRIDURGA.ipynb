# **File Reading**

```python
with open("hostel_bois.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()

print("Total lines:", len(lines))
```

# **Chat Parser**

The whatsapp chat is parsed line by line to extract the timestamp;
sender, and message. And also the special or invalid lines are
skipped.

```python
messages = []

for line in lines:
    line = line.strip()

    if line == "":
        continue

    try:
        timestamp, rest = line.split(" - ", 1)
        sender, message = rest.split(": ", 1)

        messages.append({
            "timestamp": timestamp,
            "sender": sender,
            "message": message
        })

    except ValueError:
        continue

print("Parsed messages:", len(messages))
```

# **Top 10 Words**

```python
stop_words = {
    "i", "the", "is", "am", "are", "was", "were",
    "a", "an", "and", "or", "to", "of",
    "in", "on", "at", "for", "from",
    "you", "your", "he", "his", "she", "her",
    "it", "this", "that", "have", "has",
    "had", "my", "me", "we", "they", "them"
}
word_count = {}

for message in messages:
    text = message["message"].lower()

    for punctuation in ".,!?;:\"'()[]{}":
        text = text.replace(punctuation, "")

    words = text.split()

    for word in words:
        if word not in stop_words:
            if word not in word_count:
                word_count[word] = 1
            else:
                word_count[word] += 1

top_words = sorted(
    word_count.items(),
    key=lambda x: x[1],
    reverse=True
)

print("TOP 10 WORDS")

for word, count in top_words[:10]:
    print(word, ":", count)
```

# **Response Speed and silent Streaks**

RESPONSE-TIME

```python
from datetime import datetime

for message in messages[:5]:
    time = datetime.strptime(
        message["timestamp"],
        "%d/%m/%y, %H:%M"
    )
    print(time)
    response_times = []

for i in range(1, len(messages)):
    current_time = datetime.strptime(
        messages[i]["timestamp"],
        "%d/%m/%y, %H:%M"
    )

    previous_time = datetime.strptime(
        messages[i-1]["timestamp"],
        "%d/%m/%y, %H:%M"
    )

    if messages[i]["sender"] != messages[i-1]["sender"]:
        gap = current_time - previous_time
        response_times.append(gap.total_seconds() / 60)

print("Number of responses:", len(response_times))
if len(response_times) > 0:
    average_response = sum(response_times) / len(response_times)
    print("Average response time:", round(average_response, 2), "minutes")
else:
    print("No response data available")

```

```python
from datetime import datetime, timedelta


active_dates = set()

for message in messages:
    date = datetime.strptime(
        message["timestamp"],
        "%d/%m/%y, %H:%M"
    ).date()

    active_dates.add(date)


start_date = min(active_dates)
end_date = max(active_dates)

current_date = start_date
silent_streak = 0
longest_silent_streak = 0

while current_date <= end_date:

    if current_date not in active_dates:
        silent_streak += 1

        if silent_streak > longest_silent_streak:
            longest_silent_streak = silent_streak
    else:
        silent_streak = 0

    current_date += timedelta(days=1)

print("Longest silent streak:", longest_silent_streak, "days")
```

SILENT-STREAK

```python
from datetime import datetime, timedelta

participant_dates = {}


for message in messages:
    sender = message["sender"]

    date = datetime.strptime(
        message["timestamp"],
        "%d/%m/%y, %H:%M"
    ).date()

    if sender not in participant_dates:
        participant_dates[sender] = set()

    participant_dates[sender].add(date)



start_date = min(participant_dates[next(iter(participant_dates))])
end_date = max(participant_dates[next(iter(participant_dates))])


for participant in participant_dates:

    active_dates = participant_dates[participant]

    current_date = start_date
    silent_streak = 0
    longest_silent_streak = 0

    while current_date <= end_date:

        if current_date not in active_dates:
            silent_streak += 1

            if silent_streak > longest_silent_streak:
                longest_silent_streak = silent_streak
        else:
            silent_streak = 0

        current_date += timedelta(days=1)

    print(participant, ":", longest_silent_streak, "silent days")
```

# **Personality Archetypes**

```python
from datetime import datetime

night_owl_percentage = {}

for participant in participant_list:

    total_messages = 0
    night_messages = 0

    for message in messages:

        if message["sender"] == participant:

            total_messages += 1

            time = datetime.strptime(
                message["timestamp"],
                "%d/%m/%y, %H:%M"
            )

            hour = time.hour

            if hour >= 23 or hour < 5:
                night_messages += 1

    if total_messages > 0:
        percentage = (night_messages / total_messages) * 100
    else:
        percentage = 0

    night_owl_percentage[participant] = percentage


print("NIGHT OWL PERCENTAGE")
print("---------------------")

for participant, percentage in night_owl_percentage.items():
    print(participant, ":", round(percentage, 2), "%")
```

```python
storyteller_score = {}

for participant in participant_list:

    total_messages = 0
    total_words = 0

    for message in messages:

        if message["sender"] == participant:

            total_messages += 1

            words = message["message"].split()
            total_words += len(words)

    if total_messages > 0:
        average_words = total_words / total_messages
    else:
        average_words = 0

    storyteller_score[participant] = average_words


print("STORYTELLER SCORE")
print("-----------------")

for participant, score in storyteller_score.items():
    print(participant, ":", round(score, 2), "words/message")
```

```python
drama_queen_score = {}

for participant in participant_list:

    total_valid_messages = 0
    all_caps_messages = 0

    for message in messages:

        if message["sender"] == participant:

            text = message["message"].strip()


            if len(text) < 3:
                continue

            total_valid_messages += 1

            if text.isupper():
                all_caps_messages += 1

    if total_valid_messages > 0:
        percentage = (all_caps_messages / total_valid_messages) * 100
    else:
        percentage = 0

    drama_queen_score[participant] = percentage


print("DRAMA QUEEN SCORE")
print("------------------")

for participant, percentage in drama_queen_score.items():
    print(participant, ":", round(percentage, 2), "%")
```

```python
exclamation_count = {}

for participant in participant_list:

    count = 0

    for message in messages:

        if message["sender"] == participant:

            if message["message"].count("!") >= 2:
                count += 1

    exclamation_count[participant] = count


print("DRAMA QUEEN - EXCLAMATION MARK CHECK")
print("--------------------------------------")

for participant, count in exclamation_count.items():
    print(participant, ":", count, "messages with 2+ !")
```

```python
from datetime import datetime

total_days = len(set(
    datetime.strptime(
        message["timestamp"],
        "%d/%m/%y, %H:%M"
    ).date()
    for message in messages
))

ghost_score = {}

for participant in participant_list:

    active_dates = set()

    for message in messages:

        if message["sender"] == participant:

            date = datetime.strptime(
                message["timestamp"],
                "%d/%m/%y, %H:%M"
            ).date()

            active_dates.add(date)

    silent_days = total_days - len(active_dates)

    silent_percentage = (silent_days / total_days) * 100

    ghost_score[participant] = silent_percentage


print("GHOST SCORE")
print("-----------")

for participant, percentage in ghost_score.items():
    print(
        participant,
        ":",
        round(percentage, 2),
        "% silent days"
    )
```

```python
comedian_score = {}

comedy_words = ["lol", "lmao", "haha", "rofl", "lmfao"]

for participant in participant_list:

    total_messages = 0
    funny_messages = 0

    for message in messages:

        if message["sender"] == participant:

            total_messages += 1

            text = message["message"].lower()

            for word in comedy_words:
                if word in text:
                    funny_messages += 1
                    break

    if total_messages > 0:
        percentage = (funny_messages / total_messages) * 100
    else:
        percentage = 0

    comedian_score[participant] = percentage


print("COMEDIAN SCORE")
print("--------------")

for participant, percentage in comedian_score.items():
    print(
        participant,
        ":",
        round(percentage, 2),
        "%"
    )
```

```python
question_master_score = {}

for participant in participant_list:

    total_messages = 0
    question_messages = 0

    for message in messages:

        if message["sender"] == participant:

            total_messages += 1

            text = message["message"].strip()

            if text.endswith("?"):
                question_messages += 1

    if total_messages > 0:
        percentage = (question_messages / total_messages) * 100
    else:
        percentage = 0

    question_master_score[participant] = percentage


print("QUESTION MASTER SCORE")
print("---------------------")

for participant, percentage in question_master_score.items():
    print(
        participant,
        ":",
        round(percentage, 2),
        "%"
    )
```

```python
spammer_score = {}

for participant in participant_list:

    bursts = []
    current_burst = 0

    for message in messages:

        if message["sender"] == participant:
            current_burst += 1

        else:
            if current_burst > 0:
                bursts.append(current_burst)
                current_burst = 0


    if current_burst > 0:
        bursts.append(current_burst)

    if len(bursts) > 0:
        average_burst = sum(bursts) / len(bursts)
    else:
        average_burst = 0

    spammer_score[participant] = average_burst


print("SPAMMER SCORE")
print("-------------")

for participant, score in spammer_score.items():
    print(
        participant,
        ":",
        round(score, 2),
        "messages/burst"
    )
```

```python
group_mom_score = {}

caring_keywords = [
    "okay",
    "safe",
    "eat",
    "sleep",
    "take care",
    "are you",
    "please",
    "reminder",
    "drink water",
    "don't forget"
]

for participant in participant_list:

    caring_count = 0

    for message in messages:

        if message["sender"] == participant:

            text = message["message"].lower()

            for keyword in caring_keywords:

                if keyword in text:
                    caring_count += 1


    group_mom_score[participant] = caring_count


print("GROUP MOM SCORE")
print("----------------")

for participant, score in group_mom_score.items():
    print(
        participant,
        ":",
        score,
        "caring keywords"
    )
```

```python
final_archetypes = {}

for participant in participant_list:

    if spammer_score[participant] > 3:
        archetype = "Spammer"

    elif participant == max(
        group_mom_score,
        key=group_mom_score.get
    ):
        archetype = "Group Mom"

    elif night_owl_percentage[participant] > 60:
        archetype = "Night Owl"

    elif storyteller_score[participant] > 30:
        archetype = "Storyteller"

    elif drama_queen_score[participant] > 30:
        archetype = "Drama Queen"

    elif ghost_score[participant] > 60:
        archetype = "Ghost"

    elif comedian_score[participant] == max(
        comedian_score.values()
    ):
        archetype = "Comedian"

    elif question_master_score[participant] > 25:
        archetype = "Question Master"

    else:
        archetype = "No Archetype"

    final_archetypes[participant] = archetype


print("FINAL ARCHETYPE ASSIGNMENT")
print("--------------------------")

for participant, archetype in final_archetypes.items():
    print(participant, "→", archetype)
```
